```python
import numpy as np
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
final = pd.DataFrame()
for j in range(1,231):
  headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
  webpage = requests.get('https://apna.co/jobs/jobs-in-lucknow?page={}'.format(j),headers=headers).text
  soup = BeautifulSoup(webpage,'lxml')
  company = soup.find_all('a', class_ = "block rounded-xl border border-[#E8E7EA] bg-white p-3 transition-shadow hover:shadow-md mx-3 md:mx-0")
  name = []
  cname = []
  place = []
  officetype = []
  time = []

  for i in company:
      title = i.find('h2')
      companyname = i.find('span',class_ = "text-xs text-secondary-text block min-w-0 truncate")
      where = i.find('span',class_ = "text-sm text-secondary-text flex-1 truncate")
      jobtype = i.find_all('span',class_ = "text-sm text-primary-text whitespace-nowrap text-secondary-text")

      if jobtype[0] is not None:
          officetype.append(jobtype[0].get_text(strip=True))
      else:
          officetype.append('Unavailable')
      if jobtype[1] is not None:
          time.append(jobtype[1].get_text(strip=True))
      else:
          time.append('Unavailable')
      if where is not None:
          place.append(where.get_text(strip=True))
      else:
          place.append('Unavailable')
      if title is not None:
          name.append(title.get_text(strip=True))
      else:
          name.append('Unavailable')
      if companyname is not None:
          cname.append(companyname.get_text(strip=True))
      else:
          cname.append('Unavailable')


  df = pd.DataFrame({'Company_Name' : cname, 'Job' : name, 'Place' : place, 'Work_Type' : officetype, 'Job_Type' : time})
  final = pd.concat([final,df],ignore_index=True)
```

```python
final.to_csv('apna_jobs_lucknow.csv', index=False)
```
